# Задача об урнах

> В одной урне 15 белых и 5 чёрных шаров, в другой — 6 белых и 1 чёрный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны, — белый. Решить классически и смоделировать.

## Классическое решение


```python
w_u1, b_u1, take_cnt = 15, 5, 3
u1_total = w_u1 + b_u1
w_u2, u2_total = 6, 10

full_prob = 0.0
for k in range(take_cnt + 1):
    prob_k = sc.comb(w_u1, k) * sc.comb(b_u1, take_cnt - k) / sc.comb(u1_total, take_cnt)
    cond_p = (w_u2 + k) / u2_total
    full_prob += prob_k * cond_p

round(full_prob, 4)
```

### Моделирование (Монте-Карло)

Проведём $N = 1\,000\,000$ экспериментов и оценим вероятность.

```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры моделирования
N_TRIALS = 100_0000
theoretical_p = 33 / 40  # 0.825
whites_transferred = np.random.hypergeometric(ngood=15, nbad=5, nsample=3, size=N_TRIALS)
p_draw_white = (6 + whites_transferred) / 10.0
drawn_white = np.random.rand(N_TRIALS) < p_draw_white
cumulative_successes = np.cumsum(drawn_white)
trials_indices = np.arange(1, N_TRIALS + 1)
cumulative_frequencies = cumulative_successes / trials_indices

sample_indices = np.unique(np.geomspace(10, N_TRIALS, num=2000).astype(int)) - 1
x_vals = trials_indices[sample_indices]
y_vals = cumulative_frequencies[sample_indices]
plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label="Эмпирическая частота", color="#1f77b4", linewidth=1.5)
plt.axhline(theoretical_p, color="red", linestyle="--", linewidth=1.5, 
            label=f"Теоретическая вероятность ({theoretical_p:.4f})")

plt.xscale("log") 
plt.xlabel("Число испытаний (N, логарифмическая шкала)", fontsize=11)
plt.ylabel("Относительная частота", fontsize=11)
plt.title("Сходимость эмпирической частоты к теоретической (Закон больших чисел)", fontsize=12)
plt.grid(True, which="both", ls=":", alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```
